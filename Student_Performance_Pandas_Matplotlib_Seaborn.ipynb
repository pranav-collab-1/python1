```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline

sns.set_theme(style="whitegrid")
```

 1. Load the Dataset

```python
# The UCI Student Performance CSV uses ';' as the separator
df = pd.read_csv("student-mat.csv", sep=";")

df.head()
```

```python
# Display the last five rows
df.tail()
```

```python
# Number of rows and columns
print("Shape:", df.shape)
```

```python
# Dataset information
df.info()
```

```python
# Statistical summary
df.describe()
```

 2. Selecting and Indexing Data

```python
# Select one column
df["G3"].head()
```

```python
# Select multiple columns
df[["sex", "age", "studytime", "absences", "G1", "G2", "G3"]].head()
```

```python
# Select rows using iloc
df.iloc[0:5, 0:8]
```

3. Filtering Data

```python
# Students with final grade greater than 15
high_score_students = df[df["G3"] > 15]
high_score_students.head()
```

```python
# Students with fewer than 5 absences
low_absence_students = df[df["absences"] < 5]
low_absence_students.head()
```

```python
# Students who studied more than 2 hours and scored at least 15
filtered_students = df[(df["studytime"] > 2) & (df["G3"] >= 15)]
filtered_students.head()
```

 4. Creating New Columns

```python
# Average of the three period grades
df["Average_Grade"] = df[["G1", "G2", "G3"]].mean(axis=1)

df[["G1", "G2", "G3", "Average_Grade"]].head()
```

```python
# Grade improvement from G1 to G3
df["Grade_Improvement"] = df["G3"] - df["G1"]

df[["G1", "G3", "Grade_Improvement"]].head()
```

```python
# Create a performance category using apply() and lambda
def performance_category(grade):
    if grade >= 15:
        return "Excellent"
    elif grade >= 10:
        return "Good"
    else:
        return "Needs Improvement"

df["Performance_Category"] = df["G3"].apply(performance_category)

df[["G3", "Performance_Category"]].head(10)
```

## 5. Missing Values and Data Cleaning

```python
# Count missing values in each column
df.isnull().sum()
```

```python
# Total number of missing values
print("Total missing values:", df.isnull().sum().sum())
```

```python
# Example of filling missing numerical values
# (This does not change the original data because the result is stored separately.)
df_filled = df.copy()
df_filled["G3"] = df_filled["G3"].fillna(df_filled["G3"].mean())

df_filled["G3"].isnull().sum()
```

```python
# Forward fill and backward fill examples
df_ffill = df.copy()
df_ffill = df_ffill.ffill()

df_bfill = df.copy()
df_bfill = df_bfill.bfill()

print("Forward-fill completed:", df_ffill.isnull().sum().sum(), "missing values remain")
print("Backward-fill completed:", df_bfill.isnull().sum().sum(), "missing values remain")
```

```python
# Drop rows containing missing values
df_dropna = df.dropna()

print("Original rows:", len(df))
print("Rows after dropna():", len(df_dropna))
```

 6. Grouping and Aggregation

```python
# Average final grade by gender
df.groupby("sex")["G3"].mean()
```

```python
# Average final grade by school
df.groupby("school")["G3"].mean()
```

```python
# Average final grade by study time
df.groupby("studytime")["G3"].mean().sort_values(ascending=False)
```

```python
# Multiple statistics by gender
df.groupby("sex").agg(
    Average_G3=("G3", "mean"),
    Maximum_G3=("G3", "max"),
    Minimum_G3=("G3", "min"),
    Average_Absences=("absences", "mean")
)
```

7. Sorting

```python
# Sort students by final grade
df.sort_values(by="G3", ascending=False)[
    ["school", "sex", "studytime", "absences", "G3"]
].head(10)
```

```python
# Sort by absences from highest to lowest
df.sort_values(by="absences", ascending=False)[
    ["sex", "absences", "G3"]
].head(10)
```

 8. Pivot Tables

```python
# Average final grade by gender and study time
pivot_table = pd.pivot_table(
    df,
    values="G3",
    index="sex",
    columns="studytime",
    aggfunc="mean"
)

pivot_table
```

```python
# Average final grade by school and gender
pd.pivot_table(
    df,
    values="G3",
    index="school",
    columns="sex",
    aggfunc="mean"
)
```

 9. Exporting Data

```python
# Save the processed dataset
df.to_csv("student_performance_processed.csv", index=False)

print("Processed dataset saved as student_performance_processed.csv")
```

# Matplotlib Visualizations

10. Line Plot — Absences vs Final Grade

```python
plt.figure(figsize=(8, 5))
sorted_df = df.sort_values("absences")
plt.plot(sorted_df["absences"], sorted_df["G3"], marker="o", linestyle="-", alpha=0.5)
plt.xlabel("Absences")
plt.ylabel("Final Grade (G3)")
plt.title("Absences vs Final Grade")
plt.show()
```

 11. Bar Chart — Average Final Grade by Gender

```python
avg_gender_grade = df.groupby("sex")["G3"].mean()

plt.figure(figsize=(7, 5))
plt.bar(avg_gender_grade.index, avg_gender_grade.values)
plt.xlabel("Gender")
plt.ylabel("Average Final Grade")
plt.title("Average Final Grade by Gender")
plt.show()
```

12. Grouped Bar Chart — Average Grades by Study Time and Gender

```python
grouped = df.groupby(["studytime", "sex"])["G3"].mean().unstack()

ax = grouped.plot(kind="bar", figsize=(9, 5))
ax.set_xlabel("Study Time")
ax.set_ylabel("Average Final Grade")
ax.set_title("Average Final Grade by Study Time and Gender")
plt.xticks(rotation=0)
plt.show()
```

 13. Scatter Plot — Absences vs Final Grade

```python
plt.figure(figsize=(8, 5))
plt.scatter(df["absences"], df["G3"], alpha=0.6)
plt.xlabel("Absences")
plt.ylabel("Final Grade (G3)")
plt.title("Absences vs Final Grade")
plt.show()
```

14. Histogram — Final Grade

15. Histogram — Final Grade by Gender
